### Working With File Paths

When working with files in Python, handling file paths correctly is crucial to ensure your code works across different operating systems and environments. Python provides several modules and functions for working with file paths effectively.

### Using the os module

```python
import os
cwd=os.getcwd()
print(f"Current working directory is {cwd}")
```

### Create a new directory

```python
new_directory = "package"
os.mkdir(new_directory)
print(f"Directory '{new_directory}' created")
```

### Listing Files and Directories

```python
items=os.listdir('.')  ## . basically means we are specifically looking for this folder
print(items)
```

```python
print(os.listdir("../"))       # Parent folder
```

```python
print(os.listdir("../Custom_Package"))   # Custom_Package folder in the parent
```

```python
print(os.listdir("package"))    # package folder in the current directory
```

### Joining Paths

```python
dir_name="folder"
file_name="file.txt"
full_path=os.path.join(dir_name,file_name)
print(full_path)
```

```python
dir_name="folder"
file_name="file.txt"
full_path=os.path.join(os.getcwd(),dir_name,file_name)
print(full_path)
```

### Check whether the file exists or not

```python
path='example1.txt'
if os.path.exists(path):
    print(f"The path '{path}' exists")
else:
    print(f"The path '{path}' does not exists")
```

### Checking if a Path is a File or Directory

```python
import os

path = 'package'
if os.path.isfile(path):
    print(f"The path '{path}' is a file.")
elif os.path.isdir(path):
    print(f"The path '{path}' is a directory.")
else:
    print(f"The path '{path}' is neither a file nor a directory.")
```

```python
import os

path = 'example.bin'
if os.path.isfile(path):
    print(f"The path '{path}' is a file.")
elif os.path.isdir(path):
    print(f"The path '{path}' is a directory.")
else:
    print(f"The path '{path}' is neither a file nor a directory.")
```

### Getting the absolute path

```python
relative_path='example.txt'
absolute_path=os.path.abspath(relative_path)
print(absolute_path)
```
