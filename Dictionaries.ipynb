# DICTIONARIES

Dictionaries are unordered collection of items. They store data in key-value pairs. Keys must be unique and immutable (e.g. Strings, Numbers, or Tuples), while the values can be of any type.

### Creating a Dictionary

```python
empty_dict = {}
print(type(empty_dict))
```

```python
dict1 = dict()
dict1
```

```python
student = {"name" : "Prafulla", "age" : 21 , "grade" : 9}
print(student)
```

```python
print(type(student))
```

### Only single key is allowed

```python
student = {"name" : "Prafulla" , "name" : "Omkesh" , "age" : 21}
print(student)  ## Only name is accepted and the next one usually overrides the existing name
```

### Accessing Dictionary Elements

```python
stud = {"name" : "Prafulla", "age" : 21, "grade" : "A"}
print(stud)
```

```python
## Accessing Dictionary Elements
print(stud['grade'])
print(stud['age'])
## print(stud['lastname']) --> gives an error 
```

```python
## Accessing using get() method
print(stud.get('grade'))
print(stud.get('age'))
print(stud.get('lastname')) ## doesnt give error rather returns None if no match of key is found
```

```python
print(stud.get('lastname', "Not available"))  ## added custom message instead of displaying None
```

### Modifying Dictionary Elements

Dictionary are mutable, so you can add, update or delete elements
But, only the values can be modified, 
you cannot modify a key in place. Instead, you replace it by deleting the old key and adding a new one.

```python
print(student)
```

```python
student["age"] = 103
print(student)
```

```python
student["address"] = "Pune"
print(student)
```

### Deleting key and value pair

```python
del student['age']  ## Delete key and value
print(student) 
```

### Dictionary Methods

```python
## 1. Get all the keys
print(student.keys())

## 2. Get all the values
print(student.values())

## 3. Get all the key-value pairs
print(student.items())
```

### Shallow Copy 

A shallow copy in Python creates a new outer object, but does not create copies of the nested objects.


### Problem :

```python
student
```

```python
student_copy = student
```

```python
print(student)
print(student_copy)
```

```python
student["name"] = "Prafulla"
print(student)
print(student_copy)
```

### Solution :

```python
student_copy1 = student.copy()  ## Shallow Copy
```

```python
print(student)
print(student_copy1)
```

```python
student["name"] = "Samruddhi"
print(student)
print(student_copy1)
```

### Iterating Over Dictionaries

```python
for keys in student.keys():
    print(keys)
```

```python
for values in student.values():
    print(values)
```

```python
for key,value in student.items():
    print(f"{key}:{value}")
```

### Nested Dictionaries

```python
students = {
    "student1" : {"name":"Prafulla", "age":"21"},
    "student2" : {"name":"Samruddhi", "age":20}
}
print(students)
```

Access nested dictionary elements

```python
print(students["student1"]["name"])
print(students["student1"]["age"])
```

### Iterating over Nested Dictionaries

```python
for student_id, student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

### Dictionary Comprehension

```python
squares = {x:x**2 for x in range(11)}
print(squares)
```

### Conditional Doctionary Comprehension

```python
even_squares = {x: x**2 for x in range(11) if x%2==0}
print(even_squares)
```

### Use a dictionary to count the frequency of elements in list

```python
numbers = [1,2,2,3,3,3,4,4,4,4]
frequency = {}

for number in numbers:
    if number in frequency:
        frequency[number] += 1 
    else:
        frequency[number] = 1

print(frequency)
```

### Merge two dictionaries into one

```python
dict1 = {"a" : 1, "b" : 2}
dict2 = {"b" : 3, "c" : 4}
merged_dict = {**dict1, **dict2}
print(merged_dict)
```
