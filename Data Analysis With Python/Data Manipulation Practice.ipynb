```python
df = df.rename(columns= {"Date" : "Formatted_date"})
```

```python
df.head()
```

```python
df['Formatted_date'] = df['Formatted_date'].astype(object)
```

```python
df.head()
```

```python
df.dtypes
```

```python
df
```

```python
df['Sales'] = df['Sales'].fillna(df['Sales'].mean()).astype(int)
```

```python
df.head()
```

```python
df
```

```python
df['Sales_Lambda'] = df['Sales'].apply(lambda x:x**2)
```

```python
df
```

```python
grouped_mean = df.groupby(['Product','Region'])['Sales'].mean()
print(grouped_mean)
```

```python
grouped_sum = df.groupby(['Product','Region'])['Sales'].sum()
print(grouped_sum)
```

```python
grouped_agg = df.groupby(['Product','Region'])['Sales'].agg(['mean','sum','count'])
print(grouped_agg)
```
