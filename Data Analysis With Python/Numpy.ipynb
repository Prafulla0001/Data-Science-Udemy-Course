### Numpy
NumPy is a fundamental library for scientific computing in Python. It provides support for arrays and matrices, along with a collection of mathematical functions to operate on these data structures. In this lesson, we will cover the basics of NumPy, focusing on arrays and vectorized operations.

#### NumPy is widely used in:
- Data Science and Data Analytics.
- Machine Learning and Artificial Intelligence.
- Scientific and statistical computing.
- Image processing.
- Mathematical and numerical operations.

#### Features of NumPy
1. Provides powerful multidimensional arrays.
2. Supports fast mathematical operations.
3. Allows element-wise operations using vectorization.
4. Provides built-in statistical functions.
5. Supports array slicing and indexing.
6. Uses less memory than equivalent Python lists for many numerical workloads.

### 2-D array

```python
arr4 = np.array([[1,2,3,4,5],[2,3,4,5,6]])
print(arr4)
print(arr4.shape)
```

Python supports jagged arrays natively through nested lists, whereas NumPy is designed primarily for regular multidimensional arrays.

```python
try:
    arr5 = np.array([[1,2,3,4,5],[2,3,4,5,6,7,8]])  ### This throws a ValueError because NumPy arrays are primarily designed for regular arrays.
    print(arr5)
    print(arr5.shape)
except:
    pass
```

### Create Arrays using other in-built functions

#### 1. arange() --> The np.arange() function generates evenly spaced values within a specified range.

```python
np.arange(0,10,2)
```

```python
arr1 = np.arange(1,10,2).reshape(5,1)
print(arr1)
```

#### 2. ones() --> The np.ones() function creates an array filled with ones.

```python
np.ones((3,4))
```

#### 3. eye() --> An identity matrix is a square matrix in which all diagonal elements are 1 and all other elements are 0.

```python
np.eye(3)
```

### Attributes of Numpy Array

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])

print("Array:\n", arr)
print("Shape:", arr.shape)  # Output: (2, 3)
print("Number of dimensions:", arr.ndim)  # Output: 2
print("Size (number of elements):", arr.size)  # Output: 6
print("Data type:", arr.dtype)  # Output: int64 (may vary based on platform)
print("Item size (in bytes):", arr.itemsize)  # Output: 8 (may vary based on platform)
```

### Numpy Vectorized Operation

```python
arr1=np.array([1,2,3,4,5])
arr2=np.array([10,20,30,40,50])

### Element Wise addition
print("Addition:", arr1+arr2)

## Element Wise Subtraction
print("Subtraction:", arr1-arr2)

# Element-wise multiplication
print("Multiplication:", arr1 * arr2)

# Element-wise division
print("Division:", arr1 / arr2)
```

### Universal Functions

```python
arr=np.array([2,3,4,5,6])

## square root
print("Square root:",np.sqrt(arr))

## Exponential
print("Exponent values:",np.exp(arr))

## Sine
print("Sine values:",np.sin(arr))

## natural log
print("Logarithmic values (e to what power equals 2?):",np.log(arr))
```
