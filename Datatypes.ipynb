### Datatypes

1. Datatypes are a classification of data which tell the compiler or interpreter how the programmer intends to use the data.
2. They determine the type of operations that can be performed on the data, the values that the data can take, and the amount of memory needed to store the data.

```python
# Integer
a = 42
type(a)
```

```python
# Floating-point
b = 3.14
print(type(b))
```

```python
name = "Prafulla"
type(name)
```

```python
type(True)
```
