## Importing Modules in Python: Modules and Packages
In Python, modules and packages help organize and reuse code. Here's a comprehensive guide on how to import them.

```python
import math
math.sqrt(16)
```

### Specific Import

```python
from math import sqrt,pi
print(sqrt(16))
print(sqrt(25))
print(pi)
```

### Provide alias for package so that rather than using entire package name everytime, we can refer it using alias-name

```python
import numpy as np
np.array([1,2,3,4])
```

### Import all functionalities from math package

```python
from math import *
print(sqrt(16))
print(pi)
```

### Importing from Custom or User - Defined Packages

```python
from Custom_Package.maths import addition
addition(2,3)
```

```python
from Custom_Package import maths
maths.subtraction(2,3)
```

### Compulsion is there to create the __init__.py file in each packages so to specify that it is package
