### Understanding Exceptions

Exception handling in Python allows you to handle errors gracefully and take corrective actions without stopping the execution of the program. This lesson will cover the basics of exceptions, including how to use try, except, else, and finally blocks.

### What Are Exceptions?
Exceptions are events that disrupt the normal flow of a program. They occur when an error is encountered during program execution. Common exceptions include:

1. ZeroDivisionError: Dividing by zero.
2. FileNotFoundError: File not found.
3. ValueError: Invalid value.
4. TypeError: Invalid type.

```python
try:
    result=1/2
    a=b
except ZeroDivisionError as ex:
    print(ex)
    print("Please enter the denominator greater than 0")
except Exception as ex1:
    print(ex1)
    print('Main exception got caught here')
```

```python
try:
    num=int(input("Enter a number"))
    result=10/num
except ValueError:  ### if we try to divide string by number
    print("This is not a valid number")
except ZeroDivisionError:  ### denominator is zero
    print("enter denominator greater than 0")
except Exception as ex:
    print(ex)
```

### Try, Except, Else block

Python's else block in exception handling is not equivalent to Java's finally block.

Python's else is similar to code that executes only when no exception occurs, whereas Java's finally executes whether an exception occurs or not (with a few exceptional cases).

```python
try:
    num=int(input("Enter a number:"))
    result=10/num
except ValueError:
    print("That's not a valid number!")
except ZeroDivisionError:
    print("You can't divide by zero!")
except Exception as ex:
    print(ex)
else:   ## not same as finally block
    print(f"the result is {result}")
```

try 	Executes the code being tested

except	Executes if a matching exception occurs

else	Executes if no exception occurs in try

finally	Executes during normal or exceptional control flow, including when an exception is unhandled  --> typically for cleanup and releasing the resources used in try.

### The else block is optional in Python and has no direct equivalent in this Java structure.

### Try, Except, Else and Finally

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ValueError:
    print("That's not a valid number!")
except ZeroDivisionError:
    print("You can't divide by zero!")
except Exception as ex:
    print(ex)
else:
    print(f"The result is {result}")
finally:
    print("Execution complete.")
```

### File handling using Exception Handling

```python
try:
    file = open('example1.txt', 'r')
    content = file.read()
    a = b
    print(content)

except FileNotFoundError:
    print("The file does not exist")

except Exception as ex:
    print(ex)

finally:
    if 'file' in locals() and not file.closed:
        file.close()
        print('File closed')
```

```python
try:
    file = open('example1.txt', 'r')
    content = file.read()
    a = b
    print(content)

except FileNotFoundError:
    print("The file does not exist")

except Exception as ex:
    print(ex)

finally:
    if 'file' in locals() or not file.closed:
        file.close()
        print('File closed')
```

Suppose example1.txt doesn't exist.
1. open() raises a FileNotFoundError.
2. Python immediately skips the remaining statements in the try block.
3. The matching except FileNotFoundError executes and prints the message.
4. Python then executes the finally block.
5. Since open() failed, the variable file was never assigned.
6. Your condition tries to evaluate file.closed(), causing a NameError.

The except block handles the original exception, but it does not prevent a new exception from occurring in finally.
