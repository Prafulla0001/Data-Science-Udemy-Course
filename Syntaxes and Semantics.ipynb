### Syntax and Semantics

Syntax refers to the set of rules that defines the combinations of symbols that are considered to be correctly structured programs in a language. In simple terms, syntax is about the correct arrangement of words and symbols in a code.

Semantics refers to the meaning or the interpretation of the symbols, characters, and commands in the language. It is about what the code is supposed to do when it runs.

```python
name = "Prafulla"

Name = "Mandal"

# these both are different variables because python is case sensitive
```

```python
print (name)
print (Name)
```

### Indentation

Indentation in Python is used to define the structure and hierarchy of the code. Unlike many other programming languages that use braces{} to delimit blocks of code, Python uses indentation to determine the grouping of statements. This means that all the statements within a block must be indented at the same level.



```python
i = 30
if(i == 20):
    {
        print('i is equal to 20')
    }
else:
    print('i is not equal to 20')
```
