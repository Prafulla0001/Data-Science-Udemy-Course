# SETS

Sets are a built-in datatype in Python used to store collections of unique items. They are unordered, meaning that the elements do not follow a specific order, and they do not allow duplicate elements. Sets are useful for membership tests, eliminating duplicate entries, and performing mathematical set operation like union, intersection, difference, and symmetric difference.

### Create a Set

```python
my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set = set()
print(type(my_empty_set))
```

```python
my_empty_set = set([1,2,3,5,2,4,5])  # Convert list to set and all the duplicate elements are ommitted in actual set
print(my_empty_set)
```

### Basic Set Operations 

#### 1. Add Element

```python
print(my_empty_set)
```

```python
my_empty_set.add(7)
print(my_empty_set)
```

### Adding the duplicate element wont affect anything as the set only adds the unique elements

### 2. Remove Element

```python
print(my_empty_set)
my_empty_set.remove(4)
print(my_empty_set)
```

```python
## Removing the element that is not present in the Set

my_empty_set.remove(10)
print(my_empty_set)
```

```python
## Try to remove element even if the element is not present but not throw an error

my_empty_set.discard(10)
print(my_empty_set)
```

```python
my_empty_set.discard(1)
print(my_empty_set)
```

```python
## Pop method

removed_element = my_empty_set.pop()
print(removed_element)
print(my_empty_set)
```

```python
## Clear All elements

my_empty_set.clear()
print(my_empty_set)
```

### Set Membership Test

### In for checking and Not in for checking inversely

```python
print(my_set)
```

```python
print(3 in my_set)
print(10 in my_set)

print(3 not in my_set)
print(10 not in my_set)
```

### Mathematical Operations

```python
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

print(set1)
print(set2)
```

```python
## Union

union_set = set1.union(set2)
print(union_set)
```

```python
## Intersection

intersection_set = set1.intersection(set2)
print(intersection_set)
```

```python
## Intersection_Update --> Updates the common elements from both sets into 1st set

set1.intersection_update(set2)
print(set1)
```

```python
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}
```

```python
## Difference

difference_set = set1.difference(set2)
print(difference_set)
```

```python
## Difference_Update --> Updates the difference in first set

set1.difference_update(set2)
print(set1)
```

```python
## Symmetric Difference --> returns all unique elements except the common ones

set1.symmetric_difference(set2)
print(set2)
```

### Sets Methods

```python
set1 = {1,2,3,4,5}
set2 = {3,4,5,6,7}

## is subset
print(set1.issubset(set2))

## is superset
print(set1.issuperset(set2))
```

### Remove duplicates from List

```python
lst = [1,2,34,5,6,7,1,2,4,5,6]

set1 = set(lst)
print(set1)
```

### Counting Unique Words in Text

```python
text = "In this tutorial we are discussing about sets and sets usually include only unique elements"

words = text.split()

unique_words = set(words)

print(unique_words)
print(len(unique_words))
```

### Frozen Set

In Python, a frozenset is an immutable set—a set whose elements cannot be added, removed, or changed after creation.

```python
x = frozenset([1, 2, 3])
print(x)
```

```python
s = {1, 2, 3}
s.add(4)       # works

f = frozenset({1, 2, 3})
f.add(4)       # ❌ AttributeError
```

```python
a = frozenset({1, 2, 3})
b = frozenset({2, 3, 4})

print(a | b)   # frozenset({1, 2, 3, 4})
print(a & b)   # frozenset({2, 3})
```

The important advantage is that because a frozenset is immutable, it is hashable, so you can use it as a dictionary key or put it inside another set. A simple way to remember it: set = changeable; frozenset = frozen (unchangeable).

```python
d = {
    frozenset({"apple", "banana"}): "fruit group"
}
```

```python
print(d)
```
