## NAMING CONVENTIONS
1. Variable names should be descriptive
2. They must start with a letter or an '_' and contain a letter, numbers and underscores
3. Variables name are case - sensitive

```python
# Valid variable names
first_name = "Prafulla"
last_name = "Mandal"
```

```python
# #invalid variable names
# @age = 21
# 1name = "Prafulla"
# _ = 10
```

## Understanding Variable Types

Python is dynamically typed language i.e. the type of variable is decided at the runtime

```python
age = 21 #int
height = 6.1 #float
name = "Prafulla" #string
is_Student = True #boolean
```

```python
print(type(name))
```

```python
type(height)
```

## DYNAMIC TYPING

### Python allows to convert the type of variable at runtime as the program executes

```python
var = 10
print(var, type(var))

var = "Prafulla"
print(var, type(var))

var = True
print(var, type(var))
```

```python
age = input("Enter your age: ")
print(type(age))
```

```python
age = int(input("Enter your age: "))
print(type(age))
```

## Basic Calculator

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

sum = num1 + num2
diff = num1 - num2
mul = num1 * num2
div = num1 / num2

print("Sum : ", sum)
print("Difference : ", diff)
print("Product : ", mul)
print("Quotient : ", div)
```
