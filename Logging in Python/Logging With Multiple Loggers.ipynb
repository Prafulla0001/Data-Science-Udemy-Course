### Logging with Multiple Loggers
You can create multiple loggers for different parts of your application.

```python
import logging
```

### Create a Logger for Module 1

```python
logger1 = logging.getLogger("Module 1")
logger1.setLevel(logging.DEBUG)
```

### Create a Logger for Module 2

```python
logger2 = logging.getLogger("Module 2")
logger2.setLevel(logging.WARNING)
```

### Configure logging settings

```python
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)
```

### Log Messages with Different Loggers

```python
logger1.debug("This is debug message for module 1")
logger2.warning("This is a warning message for module 2")
logger2.error("This is an error message for module 2")
logger1.critical("This is a critical message for module 1")
logger2.debug("This is a debug message for module 2") ## Wont get displayed as logger 2 level is set to WARNING; Hence, only WARNING, ERROR and CRITICAL messages can be displayed
```
