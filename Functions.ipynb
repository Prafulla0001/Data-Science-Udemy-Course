# Introduction to Functions

### Definition:

A function is a block of code that performs a specific task. Functions help in organizing code, reusing code, and improving readability.

### Syntax


```python
def function_name(parameters):
    """Docstring"""  #--> Comments that make the programmer remember what the function actually does
    # Function body
    return expression
```

```python
## why functions?
num=24
if num%2==0:
    print("the number is even")
else:
    print("the number is odd")
```

```python
def even_or_odd(num):
    """This function finds even or odd"""
    if num%2==0:
        print("the number is even")
    else:
        print("the number is odd")
```

```python
## Call this function
even_or_odd(24)
```

```python
## function with multiple parameters

def add(a,b):
    return a+b

result=add(2,4)
print(result)
    
```

### Default Parameters

```python
def greet(name):
    print(f"Hello {name} Welcome To the paradise")

greet("Prafulla")
```

```python
def greet(name="Guest"):
    print(f"Hello {name} Welcome To the paradise")

greet()
```

## Variable Length Arguments

### Positional And Keywords arguments

```python
def print_numbers(*prafulla):
    for number in prafulla:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"Prafulla")
```

### Positional arguments

```python
## Positional arguments
def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"Prafulla")
```

### Keywords Arguments

```python
### Keywords Arguments

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="Prafulla",age="32",country="India")
```

### Using positional and keywords arguments in one function

```python
def print_details(*args,**kwargs):
    for val in args:
        print(f" Positional argument : {val}")
    
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,"Prafulla",name="Prafulla",age="32",country="India")  ## Postional arguments are allowed after the keyword arguments
```

### Return Statements

```python
def multiply(a,b):
    return a*b

multiply(2,3)
```

### Return Multiple Parameters

```python
def multiply(a,b):
    return a*b,a

multiply(2,3)
```
