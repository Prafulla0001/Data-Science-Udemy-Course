## Lambda Functions in Python

Lambda functions are small anonymous functions defined using the lambda keyword. They can have any number of arguments but only one expression. They are commonly used for short operations or as arguments to higher-order functions.

### Syntax

```python
lambda arguments: expression
```

```python
lambda a,b : a+b
```

```python
addition = lambda a,b : a+b
print(type(addition))
```

```python
print(addition(5,6))
```

```python
def even (num):
    if num % 2 == 0:
        return True
```

```python
even1 = lambda num : num % 2 == 0
type(even1)
```

```python
print(even1(3))
```

```python
print(even1(10))
```

```python
def multiply(a,b,c):
    return a*b*c

print(multiply(2,3,4))
```

```python
multiply1 = lambda a,b,c : a*b*c

print(multiply1(2,3,4))
```

### Map Function

```python
### map() --> applies a function to all items in a list

numbers = [1,2,3,4,5]
def square(numbers):
    for i in numbers:
        print(i**2)
        
square(numbers)
```

```python
## Instead of doing these 

map(lambda x:x**2, numbers)
```

Map object is created at that memory location.

And map ususally has a lazy-loading technique means still it is not completely initialized

So to make it initialize we need to make use of list

```python
list(map(lambda x:x**2, numbers))
```
