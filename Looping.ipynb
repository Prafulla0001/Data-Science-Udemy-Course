### Break 

Break exits the statement exits the loop prematurely

```python
for i in range(1, 5):
    if(i == 3):
        break
    print(i)
```

### Continue

Continue skips the current iteration and continues with the next

```python
for i in range(1, 5):
    if i == 3:
        continue
    print(i)
```

### Pass

Pass statement is null operation, it does nothing

```python
for i in range(1, 5):
    if i == 3:
        print("Nothing to do here")
        pass
    print(i)
```

### Nested Loops

Nested loop is a loop inside a loop

```python
for i in range(5):
    for j in range(5):
        print(i, " ", j)
```

# Prime Numbers

```python
for i in range (1, 101):
    if(i>1):
        for j in range(2,i):
            if(i%j==0):
                break
        else:
            print(i)
```
