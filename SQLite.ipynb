### SQL And SQLite
SQL (Structured Query Language) is a standard language for managing and manipulating relational databases. SQLite is a self-contained, serverless, and zero-configuration database engine that is widely used for embedded database systems. In this lesson, we will cover the basics of SQL and SQLite, including creating databases, tables, and performing various SQL operations.

```python
import sqlite3
```

### Connect to an SQLite Database

```python
connection = sqlite3.connect('example.db')
connection
```

### Creating a Cursor

```python
cursor = connection.cursor()
```

### Creating a Table and Committing the Changes

```python
cursor.execute('''CREATE TABLE IF NOT EXISTS Employees
                ( ID INT Primary Key,
                    Name TEXT NOT NULL,
                    Age Integer,
                    Department Text
                    )''')

connection.commit()
```

```python
cursor.execute('''SELECT * FROM Employees''')
```

### Insert the Data into the Table

```python
cursor.execute('''INSERT INTO Employees(ID, Name, Age, Department)
                  VALUES(01, 'Prafulla', 20, 'Computer')''')
```

### fetchall() method is used to fetch all the data that is returned by SELECT query

```python
cursor.execute('''SELECT * FROM Employees''')
row = cursor.fetchall()

for i in row : 
    print(i)
```

### Inserting some more data

```python
cursor.execute('''INSERT INTO Employees(ID, Name, Age, Department)
                  VALUES(02, 'Samruddhi', 20, 'Finance')''')

cursor.execute('''INSERT INTO Employees(ID, Name, Age, Department)
                  VALUES(03, 'Surya', 34, 'Sports')''')
```

```python
connection.commit()
```

```python
cursor.execute('''SELECT * FROM Employees''')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

### Update the data in Table

```python
cursor.execute('''UPDATE Employees
                  SET Department = 'Data Science'
                  WHERE Name = "Prafulla"''')

connection.commit()
```

```python
cursor.execute('''SELECT * FROM Employees''')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

### Delete the data from Table

```python
cursor.execute('''DELETE FROM Employees
                  WHERE ID = 03''')

connection.commit()
```

```python
cursor.execute('''SELECT * FROM Employees''')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

### CaseStudy : Working with Sales data

```python
connection = sqlite3.connect('sales_data.db')
```

```python
cursor = connection.cursor()
```

```python
cursor.execute('''
CREATE TABLE IF NOT EXISTS sales (
    id INTEGER PRIMARY KEY,
    date TEXT NOT NULL,
    product TEXT NOT NULL,
    sales INTEGER,
    region TEXT
)
''')
```

```python
sales_data = [
    ('2023-01-01', 'Product1', 100, 'North'),
    ('2023-01-02', 'Product2', 200, 'South'),
    ('2023-01-03', 'Product1', 150, 'East'),
    ('2023-01-04', 'Product3', 250, 'West'),
    ('2023-01-05', 'Product2', 300, 'North')
]

cursor.executemany('''
                   Insert into sales(date,product,sales,region)
                   values(?,?,?,?)
                    ''',sales_data)

connection.commit()
```

```python
cursor.execute('SELECT * FROM sales')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

### Close the Connection

```python
connection.close()
```

```python
# Query data from the sales table
cursor.execute('SELECT * FROM sales')
rows = cursor.fetchall()

# Print the queried data
for row in rows:
    print(row)
```
